### Case study: commitizen
Getting data from Github repo and converted to OCEL2.0 format


```python
from pystackt import *

from configToken import GITHUB_ACCESS_TOKEN

RAW = DATA_DIR / "raw"
RAW.mkdir(parents=True, exist_ok=True)
DUCKDB_PATH = RAW / "commitizen.duckdb"
SQLITE_PATH = RAW / "ocel2_commitizen.sqlite"

from pystackt import get_github_log

get_github_log(
    GITHUB_ACCESS_TOKEN= GITHUB_ACCESS_TOKEN,
    repo_owner="commitizen-tools",
    repo_name="commitizen",
    max_issues=None,
    quack_db=str(DUCKDB_PATH),
    schema="main",
)
```

```python
from pystackt import *

export_to_ocel2(
    quack_db="./commitizen.duckdb",
    schema_in="main",
    schema_out="ocel2",
    sqlite_db="SQLITE_PATH"
)
```

```python
from pystackt import *

create_statistics_views(
    quack_db="DUCKDB_PATH",
    schema_in="main",
    schema_out="statistics"
)
```
