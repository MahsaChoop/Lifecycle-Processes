### Case study: commitizen
Getting data from Github repo and converted to OCEL2.0 format


```python
from pystackt import *

from configToken import GITHUB_ACCESS_TOKEN


DUCKDB_PATH_COMMITIZEN =  "raw/commitizen.duckdb"
SQLITE_PATH_COMMITIZEN =  "raw/ocel2_commitizen.sqlite"

from pystackt import get_github_log

get_github_log(
    GITHUB_ACCESS_TOKEN= GITHUB_ACCESS_TOKEN,
    repo_owner="commitizen-tools",
    repo_name="commitizen",
    max_issues=None,
    quack_db=str(DUCKDB_PATH_COMMITIZEN),
    schema="main",
)
```

```python
from pystackt import *

export_to_ocel2(
    quack_db= DUCKDB_PATH_COMMITIZEN,
    schema_in="main",
    schema_out="ocel2",
    sqlite_db= SQLITE_PATH_COMMITIZEN
)
```

```python
from pystackt import *

create_statistics_views(
    quack_db= DUCKDB_PATH_COMMITIZEN,
    schema_in="main",
    schema_out="statistics"
)
```

```python
from pystackt import get_github_log

get_github_log(
    GITHUB_ACCESS_TOKEN= GITHUB_ACCESS_TOKEN,
    repo_owner="TauricResearch",
    repo_name="TradingAgents",
    max_issues=None,
    quack_db= DUCKDB_PATH_TRADINGAGENTS,
    schema="main",
)
```

```python
from pystackt import *

prepare_graph_data(
    quack_db= DUCKDB_PATH_TRADINGAGENTS,
    schema_in="main",
    schema_out="graph_data_prep"
)

start_visualization_app(
    quack_db= DUCKDB_PATH_TRADINGAGENTS,
    schema="graph_data_prep"
)
```
